```python
class student:
    batch = "SP24-BSE"

    def __init__(self,name,roll_no):
        self.name = name
        self.roll_no = roll_no
    def print_details(self):
        print(f"Name :{self.name}\nRoll No:{self.roll_no}\nBatch:{self.batch}")

ahmad = student("Ahmad Hassan","23")
ahmad.print_details()
```

```python
class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def speak(self):
        print("Dogs bark")

dog = Dog()
dog. speak()
```
